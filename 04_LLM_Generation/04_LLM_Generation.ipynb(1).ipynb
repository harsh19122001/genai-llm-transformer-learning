# Logits

```python
import numpy as np

# Raw scores = logits
logits = np.array([5.2, 2.1, 1.0])

print("Logits:")
print(logits)
```

# Softmax

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

probabilities = softmax(logits)

print("Probabilities:")
print(probabilities)

print("Sum:", np.sum(probabilities))
```

# Temperature

```python
def softmax_temperature(logits, temperature=1.0):
    scaled_logits = logits / temperature
    return softmax(scaled_logits)

print("Temperature = 0.5")
print(softmax_temperature(logits, 0.5))

print("\nTemperature = 1.0")
print(softmax_temperature(logits, 1.0))

print("\nTemperature = 2.0")
print(softmax_temperature(logits, 2.0))
```

# Top-k Sampling

```python
import numpy as np

# Raw scores (logits)
logits = np.array([5.0, 4.0, 3.0, 2.0, 1.0])

k = 2

# Top-k indices
top_k_indices = np.argsort(logits)[-k:][::-1]

print("Top-k indices:", top_k_indices)
print("Top-k logits:", logits[top_k_indices])
```

```python
top_k_logits = logits[top_k_indices]

top_k_probs = np.exp(top_k_logits - np.max(top_k_logits))
top_k_probs = top_k_probs / np.sum(top_k_probs)

print("Top-k probabilities:")
print(top_k_probs)
```

# Hugging Face — Real LLM Generation

```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="distilbert/distilgpt2"
)
```
